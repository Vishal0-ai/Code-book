# Cleaning the Data Using Python Only

## Description of project:-
 Cleaning the give Dumped Data using python:-
 
 Remove users with missing names

 Remove duplicate friends

 Remove inactive users
 
 Remove duplicate pages

```python
import json
# Function For cleaning
def Clean_data(data):
    # Drop users with same id's
    unique_user = {}
    for user in data["users"]:
        unique_user[user["id"]] = user
    data["users"] = list(unique_user.values())
   
    # Drop users with missing/null/empty names
    data["users"] = [user for user in data["users"] if user.get("name") and user["name"].strip()]
    
    # Drop users with null/missing "friends"
    for user in data["users"]:
        user["friends"] = list(set(user.get("friends") or []))
    
    # Keep only users who have friends or liked_pages
    data["users"] = [user for user in data["users"] if user.get("friends") or user.get("liked_pages")]
    
    
    # Drop pages with missing/null/empty names
    data["pages"] = [page for page in data["pages"] if page.get("name") and page["name"].strip()]

    # Drop pages with same id's
    unique_pages = {}
    for page in data["pages"]:
        unique_pages[page["id"]] = page
    data["pages"] = list(unique_pages.values())
    
    return data

with open("Dumped_Data.json") as f:
    data = json.load(f)

data = Clean_data(data)

with open("cleaned_data.json", "w") as f:
    json.dump(data, f, indent=4)

print("Data cleaned successfully!")
    
```
